### **Grayscale**

Lightness: $\frac{max(R,G,B)+min(R,G,B)}{2}$

Average: $\frac{R+G+B}{3}$

Luminosity: $0.2126 \cdot R + 0.7152 \cdot G + 0.0722 \cdot B$


### **Cor**

$$\begin{bmatrix}
    R \\ G \\ B
\end{bmatrix} = \begin{bmatrix}
    255 \\ 255 \\ 255
\end{bmatrix} - \begin{bmatrix}
    C \\ M \\ Y
\end{bmatrix}$$

**f)** Crie uma função para espelhar os frames de um vídeo no eixo X.

````python
def reflectFrameX(frame): #Função que espelha o frame
    (l,c,p) = frame.shape
    rf_img = np.zeros(shape=frame.shape,dtype=np.uint8)
    for i in range(l):
        for j in range(c):
            new_x = -i
            new_y = j
            rf_img[new_x, new_y] = frame[i, j]
    return rf_img

video_reflected = video.fl_image(reflectFrameX) #Aplica a função em todos os frames
video_reflected.ipython_display(width=480) #Mostra o vídeo refletido
````
